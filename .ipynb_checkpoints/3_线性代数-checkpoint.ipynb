标量由只有一个元素的张量表示

```python
import torch
x=torch.tensor([3.0])
y=torch.tensor([2.0])
x+y,x*y,x/y,x**y
```

你可以将向量视为标量值组成的列表

```python
x=torch.arange(4)
x
```

```python
x[3]
```

访问张量的长度

```python
len(x)
```

只有一个轴的张量，形状就只有一个元素

```python
x.shape
```

通过指定两个分量m和n来创建一个形状为m*n的矩阵

```python
A=torch.arange(20).reshape(5,4)
A
```

矩阵的转置，就是沿着对角线对折

```python
A.T
```

对称矩阵(symmetric matrix) A等于其转置：A=A.T

```python
B=torch.tensor([[1,2,3],[2,0,4],[3,4,5]])
B
```

```python
B==B.T
```

就像向量是标量的推广，矩阵是向量的推广一样，我们可以构建具有更多轴的数据结构

```python
X=torch.arange(24).reshape(2,3,4)
X
```

注意看上述有三个[][][]，三维矩阵

给定具有相同形状的任何两个张量，任何按元素二元运算的结果都是相同形状的张量

```python
A=torch.arange(20,dtype=torch.float32).reshape(5,4)
B=A.clone()#通过分配新内存，将A的一个副本分配给B
A,A+B
```

两个矩阵按元素的乘法称为哈达玛积,注意看，这里和矩阵的乘法不一样，就是单纯的两数相乘

```python
A*B
```

```python
a=2
X=torch.arange(24).reshape(2,3,4)
a+X,(a*X).shape#就是X中的每个元素都加上a这个标量
```

计算其元素的和

```python
x=torch.arange(4,dtype=torch.float32)
x,x.sum()
```

表示任意形状张量的元素和

```python
A=torch.arange(20*2).reshape(2,5,4)
A.shape,A.sum(),A
```

指定求和汇总张量的轴

```python
A_sum_axis0=A.sum(axis=0)#按照第一个维度求和，就是把上下两个矩阵加起来
A_sum_axis0,A_sum_axis0.shape
```

```python
A_sum_axis1=A.sum(axis=1)
A_sum_axis1,A_sum_axis1.shape
```

```python
A.sum(axis=[0,1])
```

一个与求和相关的量是平均值

```python
A=torch.arange(40,dtype=torch.float32).reshape(2,5,4)
A.mean(),A.sum()/A.numel()#这里mean的输入必须是浮点数或复数类型
```

```python
A.mean(axis=0),A.sum(axis=0)/A.shape[0]
```

计算总和或均值时保持轴数不变

```python
sum_A=A.sum(axis=1,keepdims=True)
sum_A
```

通过广播将A除以sum_A

```python
A/sum_A
```

某个轴计算A元素的累计综合

```python
A.cumsum(axis=0)
```

点积是相同位置的按元素乘积的和

```python
y=torch.ones(4,dtype=torch.float32)
x,y,torch.dot(x,y)
```

我们可以通过执行按元素乘法，然后进行求和来表示两个向量的点积

```python
torch.sum(x*y)
```

矩阵向量积Ax是一个长度为m的列向量，也要满足A:(s,t)，x:(t,)

```python
A=torch.arange(20,dtype=torch.float32).reshape(5,4)
A.shape,x.shape,torch.mv(A,x)#mv对应matrix和vector
```

矩阵乘法AB

```python
B=torch.ones(4,3)
torch.mm(A,B)
```

L2范数是向量元素平方和的平方根，norm

```python
u=torch.tensor([3.0,-4.0])
torch.norm(u)
```

L1范数是向量元素的绝对值求和

```python
torch.abs(u).sum()
```

矩阵的佛罗贝尼乌斯范数(Frobenius norm)是矩阵元素的平方和的平方根

```python
torch.norm(torch.ones(4,9))
```

总结：norm就是元素平方和的平方根
