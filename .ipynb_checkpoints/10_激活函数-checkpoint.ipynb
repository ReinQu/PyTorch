```python
import torch
import matplotlib.pyplot as plt
```

ReLU函数：给定元素x，输出该元素与0中的最大值

```python
x=torch.arange(-8.0,8.0,0.1,requires_grad=True)
y=torch.relu(x)
plt.figure(figsize=(5, 2.5))
plt.plot(x.detach(), y.detach(), label='relu(x)')
plt.legend()
plt.show()
```

```python
y.backward(torch.ones_like(x),retain_graph=True)#backward() 对非标量输出需要指定反向传播的初始梯度。retain_graph=True：反向传播后保留计算图，方便后续再用
plt.figure(figsize=(5,2.5))
plt.plot(x.detach(),x.grad,label='grad of relu')
plt.legend()
plt.show()
```

使用ReLU的原因：它的求导表现特别好，要么让参数消失，要么让参数通过；并且能缓解神经网络的梯度消失问题。

sigmoid函数：1/(1+exp(-x))

```python
y=torch.sigmoid(x)
plt.figure(figsize=(5,2.5))
plt.plot(x.detach(),y.detach(),label='sigmoid(x)')
plt.legend()
plt.show()
```

```python
x.grad.data.zero_()
y.backward(torch.ones_like(x),retain_graph=True)
plt.figure(figsize=(5,2.5))
plt.plot(x.detach(),x.grad,label='grad of sigmoid')
plt.legend()
plt.show()
```

tanh函数：(1-exp(-2x))/(1+exp(-2x))

```python
y=torch.tanh(x)
plt.figure(figsize=(5,2.5))
plt.plot(x.detach(),y.detach(),label='tanh(x)')
plt.legend()
plt.show()
```

```python
x.grad.data.zero_()
y.backward(torch.ones_like(x),retain_graph=True)
plt.figure(figsize=(5,2.5))
plt.plot(x.detach(),x.grad,label='grad of tanh')
plt.legend()
plt.show()
```
